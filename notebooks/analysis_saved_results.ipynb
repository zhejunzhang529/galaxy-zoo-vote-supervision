# Saved-result DIS analysis
This notebook reads frozen CSV/JSON only. It does not train, score images, or select a checkpoint.


```python
from pathlib import Path
import json, pandas as pd
candidates = [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]
root = None
for candidate in candidates:
    if (candidate/'results/formal_summary.json').exists():
        root = candidate
        break
    if (candidate/'round2_resnet/results/formal_summary.json').exists():
        root = candidate/'round2_resnet'
        break
assert root is not None
summary = json.loads((root/'results/formal_summary.json').read_text())
assert len(summary['records']) == 12
print('root:', root, 'records:', len(summary['records']))
pd.DataFrame(summary['paired_mean_and_sample_sd'])
```

```python
results = root/'results'
evidence = results if (results/'vote_disagreement_group_metrics.csv').exists() else results/'closeout'
groups = pd.read_csv(evidence/'vote_disagreement_group_metrics.csv')
baselines = json.loads((evidence/'scale_matched_constant_baselines.json').read_text())
assert len(groups) == 48
print('group rows:', len(groups), 'baseline scales:', sorted(baselines['baselines']))
groups[['train_size','target','seed','group','cross_entropy','brier']].head()
```

```python
extra_metrics = pd.read_csv(results/'extra_cross_sample_metrics.csv')
extra_pool = json.loads((results/'extra_cross_sample_pool_comparison.json').read_text())
assert len(extra_metrics) == 12
assert extra_pool['extra_selected']['n'] == 1000
print('extra rows:', len(extra_metrics), 'selected:', extra_pool['extra_selected']['n'])
extra_metrics[['pool','target','seed','ce','brier','accuracy']].head()
```
